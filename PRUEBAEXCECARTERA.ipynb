### Crear un informe interactivo con Streamlit

Streamlit es un framework de Python de código abierto para crear hermosas aplicaciones web para ciencia de datos y aprendizaje automático. Es muy fácil de usar y te permite crear y compartir aplicaciones rápidamente.

A continuación, te muestro un ejemplo básico de cómo podrías usar Streamlit para mostrar tu DataFrame `df_union`. Para ejecutar este código como una aplicación Streamlit, necesitarás guardarlo en un archivo `.py` (por ejemplo, `app.py`) y luego ejecutarlo desde tu terminal con `streamlit run app.py`.

```python
import streamlit as st

# La variable df_union ya está disponible en tu entorno de Colab
# Si estuvieras en un script .py, necesitarías cargarla o definirla aquí

st.set_page_config(layout="wide")

st.title('Análisis de Cartera de Productos')

st.write('Este es un informe interactivo que muestra la unión de la base actual y la base anterior.')

# Mostrar el DataFrame de forma interactiva
st.dataframe(df_union)

st.subheader('Estadísticas Descriptivas de la Cartera Total')
st.write(df_union['Total Cartera'].describe())

st.subheader('Distribución de la Semana')
week_counts = df_union['SEMANA'].value_counts()
st.bar_chart(week_counts)

# Puedes añadir más gráficos y elementos interactivos aquí
# Por ejemplo, un filtro por 'ZONA'
selected_zona = st.sidebar.selectbox('Selecciona una ZONA', df_union['ZONA'].unique())
filtered_df = df_union[df_union['ZONA'] == selected_zona]
st.subheader(f'Datos para la ZONA: {selected_zona}')
st.dataframe(filtered_df)
```

```python
# Para guardar el código de Streamlit en un archivo
%%writefile app.py

import streamlit as st
import pandas as pd

# Asegúrate de que df_union esté disponible.
# En un entorno real, cargarías el DataFrame desde un archivo o base de datos.
# Para este ejemplo en Colab, asumimos que df_union ya está definido.
# Si estás ejecutando esto como un script independiente, puedes cargarlo así:
# df_union = pd.read_csv('your_combined_data.csv') # O desde excel, etc.

# **IMPORTANTE:** En un entorno real de Streamlit fuera de Colab,
# necesitarías volver a cargar o pasar 'df_union' a tu script 'app.py'.
# Para propósitos de demostración en Colab, estamos usando la variable existente.

# Si quieres que sea un script completamente autónomo, necesitarías
# redefinir o cargar 'base_actual' y 'base_anterior' aquí y luego concatenarlas.
# Por simplicidad para esta demostración, asumimos que `df_union` ya es global o se pasa.

# Ejemplo de cómo podrías cargarlo si fuera un script independiente:
# base_actual = pd.read_excel('Prueba especialista Candidatos.xlsx', sheet_name='BASE TOTAL (Sem Actual)')
# base_anterior = pd.read_excel('Prueba especialista Candidatos.xlsx', sheet_name='BASE TOTAL (Sem Anterior)')
# base_actual["SEMANA"] = "SEMANA ACTUAL"
# base_anterior["SEMANA"] = "SEMANA ANTERIOR"
# df_union = pd.concat([base_actual, base_anterior])

# En Colab, para que este `app.py` funcione directamente después de las celdas anteriores,
# `df_union` debe ser una variable global accesible. Si no, necesitas copiar
# las definiciones de `base_actual` y `base_anterior` y la concatenación aquí.

# Para esta demostración, usaremos un DataFrame dummy si df_union no está disponible
# para que el script no falle, pero lo ideal es que uses el `df_union` real.
try:
    # Acceder a df_union definida en el entorno de Colab
    df_union = globals()['df_union']
except KeyError:
    st.warning("df_union no encontrada. Usando un DataFrame de ejemplo para la demostración.")
    data = {'col1': [1, 2, 3], 'col2': [10, 20, 30], 'SEMANA': ['SEMANA ACTUAL', 'SEMANA ANTERIOR', 'SEMANA ACTUAL'], 'ZONA': ['ZONA A', 'ZONA B', 'ZONA A'], 'Total Cartera': [1000, 2000, 1500]}
    df_union = pd.DataFrame(data)


st.set_page_config(layout="wide")

st.title('Análisis de Cartera de Productos')

st.write('Este es un informe interactivo que muestra la unión de la base actual y la base anterior.')

# Mostrar el DataFrame de forma interactiva
st.dataframe(df_union)

st.subheader('Estadísticas Descriptivas de la Cartera Total')
st.write(df_union['Total Cartera'].describe())

st.subheader('Distribución de la Semana')
week_counts = df_union['SEMANA'].value_counts()
st.bar_chart(week_counts)

# Puedes añadir más gráficos y elementos interactivos aquí
# Por ejemplo, un filtro por 'ZONA'
if 'ZONA' in df_union.columns:
    selected_zona = st.sidebar.selectbox('Selecciona una ZONA', df_union['ZONA'].unique())
    filtered_df = df_union[df_union['ZONA'] == selected_zona]
    st.subheader(f'Datos para la ZONA: {selected_zona}')
    st.dataframe(filtered_df)
else:
    st.sidebar.warning("La columna 'ZONA' no se encuentra en el DataFrame.")
```
